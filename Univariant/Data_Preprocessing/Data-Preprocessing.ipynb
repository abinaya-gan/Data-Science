```python
import pandas as pd
```

```python
dataset = pd.read_csv("car details v4.csv")
dataset
```

```python
dataset.isnull().sum()
```

```python
dataset.isna().sum()
```

```python
from Univariant import Univariant
```

```python
import numpy as np
```

```python
quan,qual = Univariant.quanqual(dataset)
```

```python
from sklearn.impute import SimpleImputer
imp = SimpleImputer(missing_values=np.nan, strategy = 'mean')
imp.fit(dataset[quan])
df = imp.transform(dataset[quan])
df = pd.DataFrame(df, columns = quan)
```

```python
imp1 = SimpleImputer(missing_values=np.nan, strategy = 'most_frequent')
imp1.fit(dataset[qual])
df1 = imp1.transform(dataset[qual])
df1 = pd.DataFrame(df1, columns = qual)
```
